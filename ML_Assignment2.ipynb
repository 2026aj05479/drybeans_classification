```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Load the dry beans dataset
df = pd.read_excel("DryBeanDataset\Dry_Bean_Dataset.xlsx")
```

```python
print("DryBean Dataset shape:", df.shape)
print("Number of samples:", df.shape[0])
print("Number of columns:", df.shape[1])
print("Number of input features:", df.shape[1] - 1)
```

```python
df.info()
```

```python
missing_values = df.isnull().sum()

print("Missing values:")
print(missing_values)
print("Total missing values:", df.isnull().sum().sum())
```

```python
#Check for duplicate rows
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# dataset contains duplicate rows. duplicated rows are removed to improve the model performs.
df_cleaned = df.drop_duplicates().copy()
```

```python
#correlated feature are dropped to check if accuary improves. with morethan 95% relation. this did not improve the accuracy. 
#df_cleaned = df_cleaned.drop(columns=['AspectRation', 'Area', 'EquivDiameter', 'ConvexArea', 'Perimeter'])
```

```python
print("DryBean dataset shape after preprocessing:", df_cleaned.shape)
print("Number of samples:", df_cleaned.shape[0])
print("Number of columns:", df_cleaned.shape[1])
print("Number of input features:", df_cleaned.shape[1] - 1)
```

```python
# check different class/kinds of dry beans
class_counts = df_cleaned["Class"].value_counts()

print(class_counts)
```

```python
df_cleaned.describe().T
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

df_cleaned["Class"] = label_encoder.fit_transform(df_cleaned["Class"])
```

```python
print("Class mapping:")

for encoded_value, class_name in enumerate(label_encoder.classes_):
    print(f"{class_name} -> {encoded_value}")
```

```python
X = df_cleaned.drop(columns=["Class"])
y = df_cleaned["Class"]
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
correlation_matrix = X.corr()

correlation_matrix
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
```

```python
print("Training class distribution:")
print(y_train.value_counts())

print("\nTest class distribution:")
print(y_test.value_counts())
```

```python
test_data = X_test.copy()
test_data["Class"] = y_test.values

test_data.to_csv(
    "test_data.csv",
    index=False
)

print("test_data.csv created successfully.")
print("Shape:", test_data.shape)
```

```python
from sklearn.metrics import accuracy_score, roc_auc_score, precision_score, recall_score, f1_score, matthews_corrcoef

def evaluate_model(model, X_test, y_test):
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)

    accuracy = accuracy_score(y_test, y_pred)
    auc = roc_auc_score(y_test, y_prob, multi_class="ovr", average="weighted")
    precision = precision_score(y_test, y_pred, average="weighted", zero_division=0)
    recall = recall_score(y_test, y_pred, average="weighted", zero_division=0)
    f1 = f1_score(y_test, y_pred, average="weighted", zero_division=0)
    mcc = matthews_corrcoef(y_test, y_pred)

    return accuracy, auc, precision, recall, f1, mcc
```

```python
# logistic Regression 
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
```

```python
lr_model = Pipeline([("scaler", StandardScaler()), ("classifier", LogisticRegression(max_iter=2000, random_state=42))])
lr_model.fit(X_train, y_train)
```

```python
lr_accuracy, lr_auc, lr_precision, lr_recall, lr_f1, lr_mcc = evaluate_model(lr_model, X_test, y_test)

print("Logistic Regression Results")
print("---------------------------")
print(f"Accuracy  : {lr_accuracy:.4f}")
print(f"AUC       : {lr_auc:.4f}")
print(f"Precision : {lr_precision:.4f}")
print(f"Recall    : {lr_recall:.4f}")
print(f"F1 Score  : {lr_f1:.4f}")
print(f"MCC       : {lr_mcc:.4f}")
```

```python
import joblib
import os

os.makedirs("model", exist_ok=True)

joblib.dump(lr_model, "model/logistic_regression.joblib")
```

```python
# Decision Tree
from sklearn.tree import DecisionTreeClassifier

dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)
```

```python
dt_accuracy, dt_auc, dt_precision, dt_recall, dt_f1, dt_mcc = evaluate_model(dt_model, X_test, y_test)

print("Decision Tree Results")
print("---------------------")
print(f"Accuracy  : {dt_accuracy:.4f}")
print(f"AUC       : {dt_auc:.4f}")
print(f"Precision : {dt_precision:.4f}")
print(f"Recall    : {dt_recall:.4f}")
print(f"F1 Score  : {dt_f1:.4f}")
print(f"MCC       : {dt_mcc:.4f}")
```

```python
joblib.dump(dt_model, "model/decision_tree.joblib")
```

```python
#K-Nearest Neighbors
from sklearn.neighbors import KNeighborsClassifier

knn_model = Pipeline([("scaler", StandardScaler()), ("classifier", KNeighborsClassifier(n_neighbors=5))])
knn_model.fit(X_train, y_train)
```

```python
knn_accuracy, knn_auc, knn_precision, knn_recall, knn_f1, knn_mcc = evaluate_model(knn_model, X_test, y_test)

print("KNN Results")
print("-----------")
print(f"Accuracy  : {knn_accuracy:.4f}")
print(f"AUC       : {knn_auc:.4f}")
print(f"Precision : {knn_precision:.4f}")
print(f"Recall    : {knn_recall:.4f}")
print(f"F1 Score  : {knn_f1:.4f}")
print(f"MCC       : {knn_mcc:.4f}")
```

```python
joblib.dump(knn_model, "model/knn.joblib")
```

```python
from sklearn.naive_bayes import GaussianNB

gnb_model = GaussianNB()
gnb_model.fit(X_train, y_train)
```

```python
gnb_accuracy, gnb_auc, gnb_precision, gnb_recall, gnb_f1, gnb_mcc = \
    evaluate_model(gnb_model, X_test, y_test)

print("Naive Bayes Results")
print("-------------------")
print(f"Accuracy  : {gnb_accuracy:.4f}")
print(f"AUC       : {gnb_auc:.4f}")
print(f"Precision : {gnb_precision:.4f}")
print(f"Recall    : {gnb_recall:.4f}")
print(f"F1 Score  : {gnb_f1:.4f}")
print(f"MCC       : {gnb_mcc:.4f}")
```

```python
joblib.dump(gnb_model, "model/naive_bayes.joblib")
```

```python
#Support Vector Machine
from sklearn.svm import SVC

svm_model = Pipeline([("scaler", StandardScaler()), ("classifier", SVC(probability=True, random_state=42))])
svm_model.fit(X_train, y_train)
```

```python
svm_accuracy, svm_auc, svm_precision, svm_recall, svm_f1, svm_mcc = evaluate_model(svm_model, X_test, y_test)

print("SVM Results")
print("-----------")
print(f"Accuracy  : {svm_accuracy:.4f}")
print(f"AUC       : {svm_auc:.4f}")
print(f"Precision : {svm_precision:.4f}")
print(f"Recall    : {svm_recall:.4f}")
print(f"F1 Score  : {svm_f1:.4f}")
print(f"MCC       : {svm_mcc:.4f}")
```

```python
joblib.dump(svm_model, "model/svm.joblib")
```
